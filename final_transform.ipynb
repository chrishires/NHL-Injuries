```python
import pandas as pd
import numpy as numpy
import os
```

```python
# Loop through the final 2025 season injury tables and aggregate them into a single df

files = [
    f for f in os.listdir('Team Data/2025 Final')
    if f.endswith('.csv')
]
team_data = []

for file in files:
    prev_row = None
    team_file = pd.read_csv(f'Team Data/2025 Final/{file}', encoding='utf-16-le', sep='\t', engine='python')
    team_file = team_file.sort_values(by=['Player', 'Game'])
    team_file['GameAdj'] = team_file['Game']
    split = file.split('_')
    team_label = split[0]
    team_file['Team'] = team_label
    team_file['Season'] = '2025/26'
    

    for idx, row in team_file.iterrows():
        if prev_row is None:
            team_data.append(row)
        else:
            new_player = row['Player'] != prev_row['Player']
            new_injury_type = row['Injury'] != prev_row['Injury']
            gameadj_jump = row['GameAdj'] - prev_row['GameAdj']

            if new_player:
                team_data.append(row)
            elif new_injury_type:
                team_data.append(row)
            elif gameadj_jump > 1:
                team_data.append(row)
        prev_row = row

final_injuries = pd.DataFrame(team_data)
```

```python
final_injuries.sort_values(by=['Game','Player'], inplace=True)
```

```python
final_injuries.to_csv('2025_injuries_final.csv', index=False)
```
